```python
import subprocess
import sys
from pathlib import Path


def find_repo_root() -> Path:
    for candidate in (Path.cwd().resolve(), *Path.cwd().resolve().parents):
        if (candidate / "pyproject.toml").is_file() and (
            candidate / "scripts" / "configure_workshop.py"
        ).is_file():
            return candidate
    raise RuntimeError(
        "Repository root not found. Open the notebook in your cloned workshop repository."
    )


REPO_ROOT = find_repo_root()
if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))

from scripts.lib.workshop_runtime import (  # noqa: E402
    WORKSHOP_ENVIRONMENT,
    get_runtime_python,
    require_current_runtime,
)

require_current_runtime(WORKSHOP_ENVIRONMENT)
print(f"Repository: {REPO_ROOT.name}")
```

```python
subscription_id = ""
resource_group_name = ""

subscription_id = subscription_id.strip()
resource_group_name = resource_group_name.strip()
if not subscription_id or not resource_group_name:
    raise ValueError("Enter the subscription ID and resource group name in this cell.")
```

```python
subprocess.run(
    [
        str(get_runtime_python(WORKSHOP_ENVIRONMENT)),
        str(REPO_ROOT / "scripts" / "configure_workshop.py"),
        "--subscription",
        subscription_id,
        "--resource-group",
        resource_group_name,
    ],
    cwd=REPO_ROOT,
    check=True,
)
```

```python
from scripts.lib.workshop_context import load_context, validate_workshop_context

context_path = REPO_ROOT / ".workshop" / "context.json"
context = validate_workshop_context(
    load_context(context_path),
    expected_subscription_id=subscription_id,
    expected_resource_group=resource_group_name,
)
print(f"Context: {context['setup_status']} (schema {context['schema_version']})")
print(f"Resource outputs: {len(context['resource_outputs'])}")
print(f"Saved: {context_path}")
```
